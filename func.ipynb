# Подпрограммы

При решении сложных задач часто бывает удобно оформить этапы этого решения (подзадачи) в виде самостоятельных фрагментов кода, которые можно вызывать по имени (одной строчкой). Такие части программы называются **подпрограммами**.

Работая с языком Python, мы неизбежно сталкиваемся с подпрограммами с самого первого дня. Фактически известные нам `print()` и `input()` являются не чем иным как подпрограммами. Каждая из них решает одну конкретную задачу (вывод или ввод данных), имеет имя и может принимать дополнительную информацию в скобках. Давайте обратим внимание на особенности их применения:

```python
print('Hello world')
```

```python
username = input()
```

Очевидная разница в использовании объясняется тем, что `print()` просто выполняет некие действия и завершается, а `input()` еще и сообщает (_возвращает_) нам информацию о результатах этих действий. Эту информацию мы можем сохранить в переменную, вывести на экран или передать другой функции, как это происходит в конструкции вида `x = int(input())`: результат работы `input()` мы передаем в `int()`, а результат `int()` сохраняем в переменной `x`.

Использование подпрограмм позволяет:
1. Избежать дублирования кода, если одни и те же действия нужно производить в разных местах программы.
2. Структурировать код за счет разбиения сложной задачи на простые понятные действия. Такой код легче понимать и тестировать.
3. Автоматически создавать копии переменных, которые мы собираемся менять в алгоритме, а потом использовать исходные значения (см. алгоритм Евклида ниже).
4. Использовать рекурсию.

Подпрограммы, просто выполняющие действия, принято называть _процедурами_, а подпрограммы, возвращающие результат, - _функциями_. В языке python и те, и другие технически являются функциями, просто процедуры возвращают значение `None` (ничего). Наша задача сегодня - научиться писать собственные процедуры и функции.

## Процедуры

Мы начнем с процедур, но все рассматриваемые ниже идеи и примеры применимы и к функциям.

Пусть нам нужно напечатать снегопад из 4 строк: на первой строке одну снежинку, на второй - две, на третьей - три, на четвертой - четыре. Оформим решение этой подзадачи как процедуру `snow()`.

```python
def snow():
    for i in range(1, 5):
        print(('❄ ' * i).center(8))

snow()
snow()
```

```python
def snow(cnt):
    for i in range(1, cnt + 1):
        print(('❄ ' * i).center(2 * cnt))

for i in range(5):
    snow(i + 1)
```

```python
def snow(cnt, char):
    char += ' '
    for i in range(1, cnt + 1):
        print((char * i).center(2 * cnt))

snow(5, '?')
snow(3, '*')
```

```python
def snow(cnt, char='❄'):
    char += ' '
    for i in range(1, cnt + 1):
        print((char * i).center(2 * cnt))

snow(5, '%')
snow(4)
snow(cnt=3, char='$')
```

```python
def gcd(a, b):
    while a != b:
        if a > b:
            a -= b
        else:
            b -= a
    return a

a = int(input())
b = int(input())
res = gcd(a, b)
print(res)
```

```python
def gcd(a, b):
    while a * b != 0:
        if a > b:
            a %= b
        else:
            b %= a
    return a + b

a = int(input())
b = int(input())
res = gcd(a, b)
print(res)
```

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

a = int(input())
b = int(input())
res = gcd(a, b)
print(res)
```
