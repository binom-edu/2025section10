# Структуры данных

## Введение

Нам знакомы три вида алгоритмов: линейные алгоритмы, ветвления и циклы. Теоретически их применение [позволяет](https://ru.wikipedia.org/wiki/%D0%A2%D0%B5%D0%BE%D1%80%D0%B5%D0%BC%D0%B0_%D0%91%D1%91%D0%BC%D0%B0_%E2%80%94_%D0%AF%D0%BA%D0%BE%D0%BF%D0%B8%D0%BD%D0%B8) решить любую задачу структурного программирования. На практике же во многих случаях важно уметь не только обрабатывать данные, но и эффективно их **хранить**. Это достигается использованием различных *структур данных*.

Представим, что метеостанция ежечасно фиксирует информацию о погоде: температуру, давление, количество осадков и т.д. Чтобы получить статистику (среднесуточное значение и др.), нужно сохранить все полученные значения, которых может оказаться несколько тысяч и даже миллионов. Не придумывать же для каждого измерения переменные!

Структуры данных позволяют объединить несколько величин одного или различных типов в некий контейнер, дать ему имя, а внутри контейнера иметь удобный доступ к отдельным элементам, обращаясь к ним, например, по номеру.

## Список

```python
a = [7, 13, 19, 6]
print(a)
print(*a)
```

```python
a = [7, 3.14, 'крокодил', print, 2 * 2 == 4]
print(a[4])
```

```python
print(a[5])
```

```python
print(a[-1])
```

```python
print(a[len(a) - 1])
```

```python
a = [0] * 5
print(a)
```

```python
a = [1, 2, 3] + [4, 5]
print(a)
```

```python
a = []
for i in range(10):
    a.append(i)
print(a)
```

```python
a = [i for i in range(10)]
print(a)
```

```python
a = [0 for i in range(5)]
print(a)
```

```python
b = ['Крокодил' for i in range(2, 5)]
print(b)
```

```python
a = [i**2 for i in range(10)]
print(a)
```

```python
b = [i for i in a if i % 10 == 6]
print(b)
```

```python
n = int(input())
a = []
for i in range(n):
    x = int(input())
    a.append(x)
print(a)
```

```python
n = int(input())
a = [int(i) for i in input().split()]
print(a)
```

```python
import random
n = 3
a = [random.randint(0, 100) for i in range(n)]
print(a)
```

```python
for i in range(n):
    print(a[i])
```

```python
for i in a:
    print(i)
```
